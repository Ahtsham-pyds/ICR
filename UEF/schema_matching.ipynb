```python
# Source - https://stackoverflow.com/a
# Posted by Adonis Gaitatzis, modified by community. See post 'Timeline' for change history
# Retrieved 2025-12-10, License - CC BY-SA 4.0

from pathlib import Path
import pandas as pd


file_path = Path("./To-Be-Shared-Requirements&Schema-V1.xlsx")
file_extension = file_path.suffix.lower()[1:]

if file_extension == 'xlsx':
    df = pd.read_excel(pd.file.read(), engine='openpyxl')
elif file_extension == 'xls':
    df = pd.read_excel(pd.file.read())
elif file_extension == 'csv':
    df = pd.read_csv(pd.file.read())
else:
    raise Exception("File not supported")
```
